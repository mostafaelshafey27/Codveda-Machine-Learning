```python
# =========================================================
# LEVEL 2 - TASK 1: LOGISTIC REGRESSION
# CUSTOMER CHURN PREDICTION
# =========================================================


# =========================================================
# 1. IMPORT LIBRARIES
# =========================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    confusion_matrix,
    classification_report,
    roc_curve,
    roc_auc_score
)


# =========================================================
# 2. LOAD TRAINING AND TESTING DATA
# =========================================================

train_df = pd.read_csv(
    "dataset/churn-bigml-80.csv"
)

test_df = pd.read_csv(
    "dataset/churn-bigml-20.csv"
)


# =========================================================
# 3. EXPLORE THE DATA
# =========================================================

print("Training Data:")
display(train_df.head())

print("\nTraining Shape:")
print(train_df.shape)

print("\nTesting Shape:")
print(test_df.shape)

print("\nDataset Information:")
train_df.info()

print("\nMissing Values:")
print(train_df.isnull().sum())

print("\nChurn Distribution:")
print(train_df["Churn"].value_counts())


# =========================================================
# 4. SEPARATE FEATURES AND TARGET
# =========================================================

X_train = train_df.drop(
    columns=["Churn"]
)

y_train = train_df["Churn"]

X_test = test_df.drop(
    columns=["Churn"]
)

y_test = test_df["Churn"]


# Convert target from False/True to 0/1
y_train = y_train.astype(int)
y_test = y_test.astype(int)


print("\nTarget Values:")
print(y_train.value_counts())


# =========================================================
# 5. IDENTIFY CATEGORICAL FEATURES
# =========================================================

categorical_features = [
    "State",
    "Area code",
    "International plan",
    "Voice mail plan"
]


# =========================================================
# 6. ONE-HOT ENCODING
# =========================================================

X_train = pd.get_dummies(
    X_train,
    columns=categorical_features,
    drop_first=True,
    dtype=int
)

X_test = pd.get_dummies(
    X_test,
    columns=categorical_features,
    drop_first=True,
    dtype=int
)


# Make sure the test data has exactly
# the same columns as the training data
X_test = X_test.reindex(
    columns=X_train.columns,
    fill_value=0
)


print("\nTraining Data After Encoding:")
display(X_train.head())

print("\nX_train Shape:")
print(X_train.shape)

print("\nX_test Shape:")
print(X_test.shape)


# =========================================================
# 7. DEFINE NUMERICAL FEATURES
# =========================================================

numerical_features = [
    "Account length",
    "Number vmail messages",
    "Total day minutes",
    "Total day calls",
    "Total day charge",
    "Total eve minutes",
    "Total eve calls",
    "Total eve charge",
    "Total night minutes",
    "Total night calls",
    "Total night charge",
    "Total intl minutes",
    "Total intl calls",
    "Total intl charge",
    "Customer service calls"
]


# =========================================================
# 8. STANDARDIZATION
# =========================================================

scaler = StandardScaler()


# Learn mean and standard deviation
# from TRAINING data only
X_train[numerical_features] = scaler.fit_transform(
    X_train[numerical_features]
)


# Transform testing data using
# the SAME scaler
X_test[numerical_features] = scaler.transform(
    X_test[numerical_features]
)


print("\nTraining Data After Scaling:")
display(X_train.head())


# =========================================================
# 9. CREATE THE LOGISTIC REGRESSION MODEL
# =========================================================

model = LogisticRegression(
    max_iter=1000
)


# =========================================================
# 10. TRAIN THE MODEL
# =========================================================

model.fit(
    X_train,
    y_train
)


# =========================================================
# 11. MAKE CLASS PREDICTIONS
# =========================================================

y_pred = model.predict(
    X_test
)


# =========================================================
# 12. PREDICT CHURN PROBABILITIES
# =========================================================

y_prob = model.predict_proba(
    X_test
)[:, 1]


# =========================================================
# 13. COMPARE ACTUAL AND PREDICTED VALUES
# =========================================================

comparison = pd.DataFrame({
    "Actual": y_test.values,
    "Predicted": y_pred,
    "Churn Probability": y_prob
})

print("\nActual vs Predicted:")
display(comparison.head(10))


# =========================================================
# 14. MODEL EVALUATION
# =========================================================

accuracy = accuracy_score(
    y_test,
    y_pred
)

precision = precision_score(
    y_test,
    y_pred
)

recall = recall_score(
    y_test,
    y_pred
)


print("\n========== MODEL EVALUATION ==========")

print("Accuracy :", accuracy)
print("Precision:", precision)
print("Recall   :", recall)


# =========================================================
# 15. CONFUSION MATRIX
# =========================================================

cm = confusion_matrix(
    y_test,
    y_pred
)

print("\nConfusion Matrix:")
print(cm)


# =========================================================
# 16. CLASSIFICATION REPORT
# =========================================================

print("\nClassification Report:")

print(
    classification_report(
        y_test,
        y_pred
    )
)


# =========================================================
# 17. ROC CURVE AND AUC
# =========================================================

fpr, tpr, thresholds = roc_curve(
    y_test,
    y_prob
)

auc = roc_auc_score(
    y_test,
    y_prob
)

print("\nAUC Score:")
print(auc)


# =========================================================
# 18. PLOT ROC CURVE
# =========================================================

plt.figure(figsize=(8, 6))

plt.plot(
    fpr,
    tpr,
    label=f"ROC Curve (AUC = {auc:.2f})"
)

# Random classifier reference line
plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--"
)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")

plt.title("ROC Curve")

plt.legend()

plt.show()


# =========================================================
# 19. MODEL COEFFICIENTS
# =========================================================

coefficients = pd.DataFrame({
    "Feature": X_train.columns,
    "Coefficient": model.coef_[0]
})


# =========================================================
# 20. CALCULATE ODDS RATIOS
# =========================================================

coefficients["Odds Ratio"] = np.exp(
    coefficients["Coefficient"]
)


# Show strongest effects first
coefficients["Absolute Coefficient"] = abs(
    coefficients["Coefficient"]
)

coefficients = coefficients.sort_values(
    by="Absolute Coefficient",
    ascending=False
)


print("\nModel Coefficients and Odds Ratios:")

display(
    coefficients[
        [
            "Feature",
            "Coefficient",
            "Odds Ratio"
        ]
    ].head(20)
)


# =========================================================
# 21. FINAL SUMMARY
# =========================================================

print("\n========== FINAL RESULTS ==========")

print("Accuracy :", accuracy)
print("Precision:", precision)
print("Recall   :", recall)
print("AUC      :", auc)

print(
    "\nLogistic Regression Task "
    "Completed Successfully."
)
```
