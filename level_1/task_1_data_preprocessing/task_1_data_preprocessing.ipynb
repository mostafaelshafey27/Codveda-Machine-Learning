```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler


# =========================================================
# 1. LOAD THE DATASET
# =========================================================

df = pd.read_csv("dataset/2) Stock Prices Data Set.csv")

print("First 5 Rows:")
display(df.head())


# =========================================================
# 2. EXPLORE THE DATASET
# =========================================================

print("\nDataset Shape:")
print(df.shape)

print("\nDataset Information:")
df.info()


# =========================================================
# 3. CHECK FOR MISSING VALUES
# =========================================================

print("\nMissing Values Before Cleaning:")
print(df.isnull().sum())


# =========================================================
# 4. DISPLAY ROWS THAT CONTAIN MISSING VALUES
# =========================================================

print("\nRows Containing Missing Values:")
display(df[df.isnull().any(axis=1)])


# =========================================================
# 5. HANDLE MISSING VALUES
# =========================================================

# The dataset is very large and contains only a few
# rows with missing values, so these rows are removed.
df = df.dropna()


# Verify that missing values were removed
print("\nMissing Values After Cleaning:")
print(df.isnull().sum())

print("\nDataset Shape After Cleaning:")
print(df.shape)


# =========================================================
# 6. IDENTIFY CATEGORICAL COLUMNS
# =========================================================

print("\nCategorical Columns:")
print(df.select_dtypes(include=["object", "string"]).columns)


# =========================================================
# 7. ENCODE THE STOCK SYMBOL
# =========================================================

label_encoder = LabelEncoder()

df["symbol"] = label_encoder.fit_transform(df["symbol"])


# =========================================================
# 8. PROCESS THE DATE COLUMN
# =========================================================

# Convert the date column from text to datetime
df["date"] = pd.to_datetime(df["date"])

# Extract useful numerical information from the date
df["year"] = df["date"].dt.year
df["month"] = df["date"].dt.month
df["day"] = df["date"].dt.day

# Remove the original date column
df = df.drop(columns=["date"])


# =========================================================
# 9. CHECK DATA AFTER ENCODING
# =========================================================

print("\nDataset After Encoding:")
display(df.head())

print("\nData Types After Encoding:")
print(df.dtypes)


# =========================================================
# 10. SPLIT DATA INTO TRAINING AND TESTING SETS
# =========================================================

train_df, test_df = train_test_split(
    df,
    test_size=0.20,
    random_state=42
)

print("\nTraining Set Shape:")
print(train_df.shape)

print("\nTesting Set Shape:")
print(test_df.shape)


# =========================================================
# 11. STANDARDIZE NUMERICAL FEATURES
# =========================================================

numerical_features = [
    "open",
    "high",
    "low",
    "close",
    "volume"
]

scaler = StandardScaler()

# Create copies before modifying the split datasets
train_df = train_df.copy()
test_df = test_df.copy()


# Learn scaling parameters from the training data
# and transform the training data
train_df[numerical_features] = scaler.fit_transform(
    train_df[numerical_features]
)


# Transform the testing data using the SAME scaler
test_df[numerical_features] = scaler.transform(
    test_df[numerical_features]
)


# =========================================================
# 12. DISPLAY FINAL RESULTS
# =========================================================

print("\nTraining Data After Scaling:")
display(train_df.head())

print("\nTesting Data After Scaling:")
display(test_df.head())


# =========================================================
# 13. FINAL CHECK
# =========================================================

print("\nFinal Training Data Information:")
train_df.info()

print("\nFinal Testing Data Information:")
test_df.info()
```
